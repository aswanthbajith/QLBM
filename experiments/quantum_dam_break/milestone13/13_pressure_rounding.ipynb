## Compare the same input with the previous circuit
A triple transfer changes the integer second moment by two while preserving the first moment and mass.

```python
o=CollisionOracle(32,omega_num=4)
b=Prior(32,omega_num=4)
f=np.zeros((1,25),dtype=np.int64)
f[0,12]=48
z,flag=o.evaluate(f)
p,_=b.evaluate(f)
print("Mass:",f.sum(),z.sum())
print("Momentum:",f@r.V,z@r.V)
print("Second moments:",p@(r.V**2),z@(r.V**2))
assert np.array_equal(z[0],r.collision(f[0],32,o.w,4)[0])
assert not flag.any()
```
