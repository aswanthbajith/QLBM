## Test actual local reversible gates
The collision retains old populations and returns an output and a rejection flag. The simulator verifies scratch cleanup and inverse execution.

```python
oracle=CollisionOracle(128,2)
f=ref.initialize(np.array([1.5,1.,0.]),np.zeros((3,2)),128)
post,reject=oracle.evaluate(f)
print(post,reject)
assert np.array_equal(post.sum(-1),f.sum(-1))
assert np.array_equal(post@C,f@C)
```
