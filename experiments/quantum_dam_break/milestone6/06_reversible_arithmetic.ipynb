# Milestone 6: Reversible arithmetic

This notebook uses executed computational-basis quantum gate simulation, not QPU data. Run from the milestone6 folder. The local flux oracle uses X/CNOT/Toffoli and uncomputes its work registers.

```python
import sys, pathlib
sys.path.insert(0,str(pathlib.Path.cwd().parent))
from milestone6.arithmetic import FluxOracle
a=FluxOracle(32,1)
key=a.pack(48,0,0)
print(a.unpack(a.evaluate(key)), len(a.gates), a.nqubits)
assert a.apply(a.apply(key),True)==key
```
