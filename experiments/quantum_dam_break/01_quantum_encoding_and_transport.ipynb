```python
p=stream_permutation(g)
assert np.allclose(apply_gates(a,p),apply_permutation(a,p))
print("Controlled-X gates:",len(list(gate_pairs(p))))
```

```python
rho=np.outer(a,a.conj())
r=relaxation_channel(g,rho,.5)
print("Trace:",np.trace(r))
```
