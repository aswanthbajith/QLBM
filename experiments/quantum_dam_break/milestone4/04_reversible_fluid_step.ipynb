This is a fixed-point gate-simulated finite-volume prototype, not D2Q9 QLBM or real quantum hardware. Run from this notebook directory; the independent reference is for validation only.

```python
import sys
from pathlib import Path
import numpy as np
root=Path.cwd()
if (root/"milestone4").is_dir(): root=root/"milestone4"
sys.path.insert(0,str(root))
from solver import StepCircuit
from run_step import run
```

```python
c=StepCircuit(6)
s=[(6,0)]*3+[(0,0)]*3
w=c.encode(s)
z=c.apply(w)
print("Qubits:",c.nqubits,"Logical gates:",c.gate_count())
print("Initial:",s,"Next:",c.decode(z))
assert c.apply(z,True)==w
assert z>>c.scratch==0
```
