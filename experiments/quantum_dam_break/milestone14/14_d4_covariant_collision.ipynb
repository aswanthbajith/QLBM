# M14: D4-covariant quantum collision
Run from this directory. The local coherent test is a sparse complex-state simulation, not a real QPU experiment.

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent if Path.cwd().name=="milestone14" else Path.cwd()))
import numpy as np
from milestone14.twirl import TwirledCollision, GROUP, action, coherent_dilation
from milestone11.reference import V
```

```
o=TwirledCollision(32,omega_num=2)
f=np.zeros((2,25),dtype=int)
f[0,12]=20;f[0,13]=4;f[1,12]=15;f[1,7]=3
y,flags=o.branches(f)
assert not flags.any()
assert np.all(y.sum(-1)==f.sum(-1)[:,None])
assert np.all(y@V==(f@V)[:,None,:])
print("Local qubits",o.nqubits,"wrapper gates",o.wrapper_gates)
print("First mean populations",y[0].mean(0).reshape(5,5))
```

```
for p in GROUP:
    z,_=o.branches(action(f,p))
    assert all(sorted(map(tuple,a))==sorted(map(tuple,b)) for a,b in zip(z,action(y,p)))
state,error=coherent_dilation(o,f,np.array([1,1j])/np.sqrt(2))
print("Sparse support",len(state),"inverse error",error)
```
